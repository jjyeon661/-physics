```python
import math
```

```python
angle=30
g=9.8
v0=30
pi=3.141592
```

```python
rad=angle*pi/180
v0x=v0*math.cos(rad)
v0y=v0*math.sin(rad)
print("v0x의 초기속도= ", v0x)
print("v0y의 초기속도=", v0y)
```

```python
h=v0y*v0y/ 2 * g
print("최고점의 높이 = ", h)
```

```python
t=v0y/g
print("최고점에 도착하는 시간 = ", t)
```

```python
s= v0*2*t
print("수평으로 이동한 최고 거리 =", s)
```

```python
vx=v0x
vy=-v0y
v=math.sqrt(vx*vx+vy*vy)
print(" 지면에 닿을때 vx= ", vx)
print(" 지면에 닿을때 vy= ", vy)
print(" 지면에 닿을때 v= ", v)
```
