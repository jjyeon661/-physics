```python
import math

# 초기 값들
v0 = 100.0          # 초기 속도 (m/s)
angle_deg = 37.0    # 발사 각도 (도)
y0 = 140.0          # 초기 높이 (m)
g = 9.8             # 중력가속도 (m/s^2)
PI = 3.141592

# 각도를 라디안으로 변환
angle_rad = angle_deg * PI / 180.0

# 수평 및 수직 초기 속도 성분
v0x = v0 * math.cos(angle_rad)
v0y = v0 * math.sin(angle_rad)

# (a) 지면에 도달하는 데 걸리는 시간 t 구하기 (이차방정식)
a = -0.5 * g
b = v0y
c = y0

dis  = b**2 - 4 * a * c
if dis < 0:
    print("실수 해가 없습니다.")
    exit()

# 양의 시간 값 선택
t1 = (-b + math.sqrt(dis)) / (2 * a)
t2 = (-b - math.sqrt(dis)) / (2 * a)
t = t1 if t1 > 0 else t2

print(f"(a) 지면에 도달하는 시간: {t:.2f} 초")

# (b) 수평 거리 x 계산
x = v0x * t
print(f"(b) 수평 거리: {x:.2f} m")

# (c) 충돌 직전 속도의 x, y 성분
vx = v0x
vy = v0y - g * t
print(f"(c) 충돌 직전 속도 성분: vx = {vx:.2f} m/s, vy = {vy:.2f} m/s")

# (d) 속도 크기 및 충돌각
v = math.sqrt(vx**2 + vy**2)
theta = math.atan2(vy, vx) * 180.0 / PI
print(f"(d) 속도 크기: {v:.2f} m/s")
print(f"(e) 충돌 각도 (수평과 이루는 각): {theta:.2f} 도")
```
